### **1. Objective**
The main objective of this project is to build a predictive model using the Simple Linear Regression algorithm of Machine Learning. This model will accurately predict the daily revenue of an ice cream parlor based on the day's temperature.

```python
# Step 1: Import necessary libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error, mean_squared_error
```

```python
# Step 2: Import data
icecream = pd.read_csv('https://github.com/ybifoundation/Dataset/raw/main/Ice%20Cream.csv')
```

```python
# Step 3: Define Target (y) and Features (X)
y = icecream['Revenue']
X = icecream[['Temperature']] # Using double brackets for 2D Array
```

```python
# Step 4: Train Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7, random_state=2529)
```

```python
# Step 5 & 6: Select Model and Train/Fit Model
model = LinearRegression()
model.fit(X_train, y_train)
print("Intercept (c):", model.intercept_)
print("Coefficient (m):", model.coef_[0])
```

```python
# Step 7: Predict Model
y_pred = model.predict(X_test)
```

```python
# Step 8: Model Evaluation / Accuracy
print("Mean Absolute Error (MAE):", mean_absolute_error(y_test, y_pred))
print("Mean Squared Error (MSE):", mean_squared_error(y_test, y_pred))
print("Mean Absolute Percentage Error (MAPE):", mean_absolute_percentage_error(y_test, y_pred))
```

### **2. Methodology**
* **Data Loading & Splitting:** The dataset was imported using `pandas`, and the Feature (X) and Target (y) variables were separated. Then, `train_test_split` was used to divide the data into Training (70%) and Testing (30%) sets.
* **Model Training & Prediction:** The `LinearRegression()` model was selected and fitted using the training data. Then, the test data was used to predict the revenue.

### **3. Challenges & Solutions**
* **Challenge:** Defining the X (Feature) variable could lead to a dimension shape issue (ValueError) as the model expects a 2D array.
* **Solution:** To send X as a 2D DataFrame, double square brackets `icecream[['Temperature']]` were used.

### **4. Conclusion**
After evaluating the model, it was observed that the Error rate is only 4.2%. This means the model is operating with approximately 95.8% accuracy. There is a strong positive correlation between environmental temperature and ice cream sales, making this model highly effective for real-world business and sales planning.
